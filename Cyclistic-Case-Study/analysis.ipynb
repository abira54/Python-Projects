#  Collect Data

```python
df_nov24 = pd.read_csv(r'divvy-tripdata - backup\202411-divvy-tripdata.csv')
df_dec24 = pd.read_csv(r'divvy-tripdata - backup\202412-divvy-tripdata.csv')
df_jan25 = pd.read_csv(r'divvy-tripdata - backup\202501-divvy-tripdata.csv')
df_feb25 = pd.read_csv(r'divvy-tripdata - backup\202502-divvy-tripdata.csv')
df_mar25 = pd.read_csv(r'divvy-tripdata - backup\202503-divvy-tripdata.csv')
df_apr25 = pd.read_csv(r'divvy-tripdata - backup\202504-divvy-tripdata.csv')
df_may25 = pd.read_csv(r'divvy-tripdata - backup\202505-divvy-tripdata.csv')
df_june25 = pd.read_csv(r'divvy-tripdata - backup\202506-divvy-tripdata.csv')
df_jul25 = pd.read_csv(r'divvy-tripdata - backup\202507-divvy-tripdata.csv')
df_aug25 = pd.read_csv(r'divvy-tripdata - backup\202508-divvy-tripdata.csv')
df_sept25 = pd.read_csv(r'divvy-tripdata - backup\202509-divvy-tripdata.csv')
df_oct25 = pd.read_csv(r'divvy-tripdata - backup\202510-divvy-tripdata.csv')
```

```python
df = pd.concat([df_nov24,df_dec24,df_jan25,df_feb25,df_mar25,df_apr25,df_may25,df_june25,df_jul25,df_aug25,df_sept25])
df
```

# Wrangle Data

```python
df.info()
```

```python
df['ride_id'] = df['ride_id'].astype('string')
df['rideable_type'] = df['rideable_type'].astype('string')
df['start_station_name'] = df['start_station_name'].astype('string')
df['start_station_id'] = df['start_station_id'].astype('string')
df['end_station_name'] = df['end_station_name'].astype('string')
df['end_station_id'] = df['end_station_id'].astype('string')
df['member_casual'] = df['member_casual'].astype('string')
```

```python
df['started_at'] = pd.to_datetime(df['started_at'])
df['ended_at'] = pd.to_datetime(df['ended_at'])
```

Add columns that list the date, month, day, and year of each ride.
This will allow us to aggregate ride data for each month, day, or year ... before completing
these operations we could only aggregate at the ride level

```python
df['date'] = df['started_at'].dt.date
df['day'] = df['started_at'].dt.day
df['month'] = df['started_at'].dt.month
df['year'] = df['started_at'].dt.year
df['weekday'] = df['started_at'].dt.day_name()
```

Add ride_length column (in seconds)

```python
df['ride_length'] = df['ended_at'] - df['started_at']
df['ride_length'] = df['ride_length'].dt.total_seconds()
df['ride_length']
```

```python
df.head()
```

Removing "bad" data
The dataframe includes a few hundred entries when bikes were taken out of docks and
checked for quality by Divvy or ride_length was negative
We will create a new version of the dataframe (v2) since data is being removed

```python
df_v2 = df[~((df['start_station_name'] == "HQ QR") | (df['ride_length']<0))].copy()
df_v2
```

# CONDUCT DESCRIPTIVE ANALYSIS

Descriptive analysis on ride_length (all figures in seconds)

```python
df_v2['ride_length'].describe()
```

Compare members and casual users

```python
df_v2.groupby('member_casual')['ride_length'].agg(['mean', 'median', 'max', 'min'])
```

See the average ride time by each day for members vs casual users

```python
df_v2.groupby(['member_casual', 'weekday'], observed= True).agg(average_duration = ('ride_length', 'mean'))
```

Fix the order of days and re-run

```python
day_order = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
# Convert 'weekday' to a Categorical type with the correct order
df_v2['weekday'] = pd.Categorical(
    df_v2['weekday'],
    categories=day_order,
    ordered=True
)
```

Now, let's run the average ride time by each day for members vs casual users

```python
df_v2.groupby(['member_casual', 'weekday'], observed= True).agg(average_duration = ('ride_length', 'mean'))
```

analyze ridership data by type and weekday

```python
rides_summary = (df_v2.groupby(['member_casual', 'weekday'], observed=True).agg(
        number_of_rides=('ride_length', 'count'),
        average_duration=('ride_length', 'mean')
        ).reset_index(
            
        ).sort_values(
        by=['member_casual', 'weekday']
        )
)
print(rides_summary)
```

## Let's visualize
The number of rides by rider type

```python
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(
    data=rides_summary,
    x='weekday',
    y='number_of_rides',
    hue='member_casual',
    order=day_order
)
plt.title('Number of Rides by Day of Week and Rider Type')
plt.xlabel('Day of Week')
plt.ylabel('Number of Rides')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Rider Type')
plt.tight_layout()
plt.show()
```

The average duration

```python
plt.figure(figsize=(10, 6))
sns.barplot(
    data=rides_summary,
    x='weekday',
    y='average_duration',
    hue='member_casual',
    order=day_order
)
plt.title('Average Ride Duration by Day of Week and Rider Type')
plt.xlabel('Day of Week')
plt.ylabel('Average Duration (Seconds)')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Rider Type')
plt.tight_layout()
plt.show()
```
